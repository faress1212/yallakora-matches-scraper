# Yallakora Matches Scraper ⚽

A web scraping project that extracts football match results (teams, score, kickoff time, TV channel, and match link) from [Yallakora.com](https://www.yallakora.com) for a given date.

> ⚠️ **Note:** Website structures can change over time. If this scraper stops finding results, inspect the page in your browser's DevTools and update the selectors in `src/scraper.py`. Always check a website's `robots.txt` and Terms of Service before scraping.


## 1. Imports

```python
from src.scraper import fetch_matches_page, parse_matches, scrape_matches_for_date
from src.cleaning import clean_matches, split_score

import pandas as pd
```

## 2. Fetching a Single Day's Matches

Enter a date in the form `MM/DD/YYYY` (e.g. `11/23/2022`, the day of Scotland vs. Morocco and Brazil vs. Haiti at the 2022 World Cup).

```python
date = input("Enter the date as MM/DD/YYYY: ")
soup = fetch_matches_page(date)
```

## 3. Parsing Match Details

We extract only *finished* matches (cards with the `item finish liItem` class), pulling out both teams, the final score, kickoff time, broadcasting channel, and a link to match details.

```python
matches = parse_matches(soup)
df = pd.DataFrame(matches)
df
```

## 4. Cleaning the Scraped Data

- Drop rows where both team names failed to parse
- Drop exact duplicates
- Split the combined `score` column (e.g. "3 - 0") into separate numeric `score_a` / `score_b` columns for easier analysis

```python
df = clean_matches(df)
df = split_score(df)
df
```

## 5. Saving the Results

```python
df.to_csv(f"../data/yallakora_matches_{date.replace('/', '-')}.csv", index=False, encoding="utf-8-sig")
print(f"Saved {len(df)} matches.")
```

## 6. Troubleshooting

If `parse_matches` returns an empty list:

1. Open the [matches page](https://www.yallakora.com/matches) for a date with finished matches, in your browser.
2. Inspect a finished match card with DevTools.
3. Confirm the class names in the `SELECTORS` dictionary in `src/scraper.py` still match (e.g. `item finish liItem`, `teams teamA`, `score`).
4. Update as needed and re-run.

Note: this scraper specifically targets **finished** matches. Live or upcoming matches use different card classes on Yallakora and aren't covered here.


## Conclusion

This project scrapes structured football match data from Yallakora for any given date, with cleaning and score-splitting for further analysis — e.g. tracking a team's results over a tournament, or building a simple results archive.
